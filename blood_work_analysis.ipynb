```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
load_dotenv()

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
with open ("blood_work.txt","r") as f:
    blood_report=f.read()
print(blood_report[:200])
```

```python
extraction_prompt = f"""
You are a medical data extraction assistant.

From the blood report below, extract ALL test values and classify each one as HIGH, LOW, or NORMAL
based on the reference ranges provided in the report.

Format your response as:
- Test Name: value | Status: HIGH/LOW/NORMAL | Reference: range

Blood Report:
{blood_report}
"""

extraction_response=llm.invoke(extraction_prompt)
extraction_values=extraction_response.text

print("stage 1 values extracted")
print(extraction_values)
```

```python
diet_prompt = f"""
You are a clinical nutritionist specializing in Indian diets.

Analyze the blood work below and respond in a very concise, simple format.

OUTPUT EXACTLY:

Health Summary:
- Write exactly 3 short sentences.

Foods to Avoid:
- 3 to 5 short bullet points.

Foods to Eat More:
- 3 to 5 short bullet points.

Rules:
- Use simple language.
- Keep every bullet under 12 words.
- Mention only foods relevant to the blood work.
- Do not add explanations or disclaimers.
- Do not add any extra sections.
- Do not repeat information.
- Do not use emojis.
- Do not use markdown tables.

Blood Work Analysis:
{extraction_values}
"""

diet_response = llm.invoke(diet_prompt)

print(diet_response.content)
```
